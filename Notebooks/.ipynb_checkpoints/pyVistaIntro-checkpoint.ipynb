```python
import numpy as np 
import pyvista as pv 

points = np.random.rand(100,3)
mesh   = pv.PolyData(points) # gives you object of PolyData type
mesh.plot(point_size=10,style='points',color='tan') # calls plot method of PolyData class to visualize the mesh
```

```python
from pyvista import examples 

mesh = examples.load_hexbeam()
cpos = [(6.20, 3.00, 7.50),
        (0.16, 0.13, 2.65),
        (-0.28, 0.94, -0.21)]

pl = pv.Plotter()
pl.add_mesh(mesh,show_edges=True,color='white') 
pl.add_points(mesh.points, color='red', point_size=20)
pl.camera_position = cpos 
pl.show()
```

```python
mesh = examples.download_bunny_coarse()

pl = pv.Plotter()
pl.add_mesh(mesh, show_edges=True, color='white')
pl.add_points(mesh.points, color='red', point_size=20)
pl.camera_position = [(0.02, 0.30, 0.73),
                      (0.02, 0.03, -0.022),
                      (-0.03, 0.94, -0.34)]
pl.show()
```
