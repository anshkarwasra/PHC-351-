```python
# cells 

mesh =  examples.load_hexbeam()

pl = pv.Plotter()
pl.add_mesh(mesh,show_edges=True,color='white')
pl.add_points(mesh.points,color='red',point_size=20)

single_cell =  mesh.extract_cells(mesh.n_cells - 1 )
pl.add_mesh(single_cell,color='yellow',edge_color='blue',line_width=5,show_edges=True)

pl.camera_position = [(6.20, 3.00, 7.50),
                      (0.16, 0.13, 2.65),
                      (-0.28, 0.94, -0.21)]
pl.show()
```

```python
import pandas as pd, glob
fs = glob.glob("data/evaluation_reports/val_sweep_*.csv")
d = pd.concat([pd.read_csv(f, dtype={"id": str}) for f in fs], ignore_index=True)
d = d[d.get("status", "ok") == "ok"] if "status" in d else d
d["rel_poor_pct"] = 100 * (d.post_poor_pct - d.pre_poor_pct) / d.pre_poor_pct
d["growth"] = d.n_elements_post / d.n_elements if "n_elements_post" in d else float("nan")
cols = [c for c in ["rel_poor_pct", "post_poor_pct", "post_min_sj", "post_severe_pct",
        "fe_iters", "growth", "flagged_elements", "modified_fraction",
        "t_correction_sec", "t_total_sec"] if c in d]
print(d.groupby(["tier", "method"])[cols].median().round(3).to_string())
for c in ["fallback", "fe_success"]:
    if c in d:
        print(c, d.groupby("method")[c].mean().round(3).to_dict())
print("n meshes per method:", d.groupby("method").size().to_dict())
```

```python
import pandas as pd, glob
d = pd.concat([pd.read_csv(f, dtype={"id": str}) for f in glob.glob("data/evaluation_reports/val_sweep_*.csv")], ignore_index=True)
print("methods:", sorted(d.method.unique()))
if "classical" in set(d.method):
    pre = d[d.method == "classical"][["id", "tier", "poor_pct", "min_sj", "n_tets"]]
else:
    b = pd.read_csv("data/evaluation_reports/benchmark_long.csv", dtype={"id": str})
    pre = b[b.method == "classical"][["id", "tier", "pre_poor_pct", "pre_min_sj", "n_elements"]].rename(
        columns={"pre_poor_pct": "poor_pct", "pre_min_sj": "min_sj", "n_elements": "n_tets"})
    print("ids matched to benchmark_long:", len(set(pre.id) & set(d.id)), "of", d.id.nunique())
pre = pre.rename(columns={"poor_pct": "pre_poor", "min_sj": "pre_min_sj", "n_tets": "pre_n"})
post = d[d.method != "classical"].merge(pre, on=["id", "tier"], how="left")
post["rel_poor_pct"] = 100 * (post.poor_pct - post.pre_poor) / post.pre_poor
post["growth"] = post.n_tets / post.pre_n
g = post.groupby(["tier", "method"]).agg(
    n=("id", "size"), rel_poor_pct=("rel_poor_pct", "median"), poor_pct=("poor_pct", "median"),
    min_sj=("min_sj", "median"), growth=("growth", "median"), moved=("moved_points", "median"),
    flagged=("flagged_cells", "median"), rejected=("rejected_moves", "median"),
    t_smooth=("t_smoothing_sec", "median")).round(3)
print(g.to_string())
```
